```python
import os
from IPython.display import Audio, display
import glob

comparison_dir = "inferece_data/saved/dataout/train"

print(f"--- Comparing from directory '{comparison_dir}' ---")

if not os.path.isdir(comparison_dir):
    print(f"Error: Comparison directory '{comparison_dir}' not found. Skipping.")
else:
    original_files = sorted(glob.glob(os.path.join(comparison_dir, "orig_audio_*.wav")))
    predicted_files = sorted(glob.glob(os.path.join(comparison_dir, "pred_audio_*.wav")))

    if not original_files:
        print(f"No 'orig_audio_*.wav' files found in '{comparison_dir}'.")
    if not predicted_files:
        print(f"No 'pred_audio_*.wav' files found in '{comparison_dir}'.")

    if original_files and predicted_files:
        print(f"Displaying all {min(len(original_files), len(predicted_files))} pairs of original and predicted audio files for comparison:")

        for i in range(min(len(original_files), len(predicted_files))):
            print(f"Original audio: {os.path.basename(original_files[i])}")
            display(Audio(original_files[i]))

            print(f"Predicted audio: {os.path.basename(predicted_files[i])}")
            display(Audio(predicted_files[i]))
            print("\n" + "-" * 20 + "\n")
    else:
        print("Could not find both original and predicted audio files to compare in this directory.")
print("-" * 50)
```
